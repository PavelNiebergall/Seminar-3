# Jupyter Notebook: Einstieg

Dieses Notebook zeigt die Grundidee von Jupyter an einem kleinen Python-Beispiel. Du kannst Text und Code mischen und die Ergebnisse direkt unter den Codezellen sehen.

## 1. Was ist Jupyter Notebook?

Ein Jupyter Notebook ist ein interaktives Dokument aus einzelnen Zellen. Markdown-Zellen enthalten Erklärungen; Codezellen führen Python aus und zeigen Ergebnisse an. Das ist nützlich für interaktive Datenanalyse, Dokumentation und Visualisierung: Rechenschritte und ihre Erklärung bleiben zusammen.

Ein Notebook arbeitet mit einem **Kernel**. Das ist der laufende Python-Prozess, der Code ausführt und Variablen zwischen Zellen speichert. Zellen lassen sich einzeln ausführen, teilen aber denselben Zustand. Daher ist die Ausführungsreihenfolge wichtig.

## 2. Installation und Start

Jupyter kannst du in einer Python-Umgebung mit `pip` installieren. Öffne ein Terminal und führe aus:

```bash
pip install notebook
jupyter notebook
```

Der Startbefehl öffnet normalerweise eine Notebook-Oberfläche im Browser. Dort kannst du ein vorhandenes Notebook öffnen oder ein neues Python-Notebook anlegen. In VS Code kannst du `.ipynb`-Dateien ebenfalls öffnen; dafür brauchst du die Python- und Jupyter-Erweiterungen.

Für das Diagramm in diesem Beispiel wird zusätzlich `matplotlib` benötigt. Installiere es in einer Codezelle in derselben Notebook-Umgebung:

```python
%pip install matplotlib
```

Nach der Installation kann ein Kernel-Neustart nötig sein. Führe dann die benötigten Codezellen erneut aus.

## 3. Codezellen erstellen und ausführen

Eine Codezelle enthält Python. Gib Code in eine Zelle ein und drücke **Shift+Enter**, um sie auszuführen. Die Ausgabe erscheint direkt darunter; zugleich springt der Fokus zur nächsten Zelle. Du kannst dieselbe Zelle später erneut ausführen.

````python
print("Hallo Jupyter!")
````

## 4. Markdown-Zellen für Dokumentation

Markdown-Zellen sind für erklärenden Text gedacht. Überschriften entstehen mit `#`, Listen mit `-` und **fetter Text** mit `**...**`. Eine Zelle lässt sich in der Notebook-Oberfläche vom Code- in den Markdown-Modus umstellen. Mit **Shift+Enter** wird Markdown formatiert angezeigt, statt als Python ausgeführt zu werden.

So kannst du festhalten, welche Frage du untersuchst und wie du ein Ergebnis interpretierst.

## 5. Python-Beispiel: zwei Zahlen addieren

Jetzt kommt eine kleine Rechnung. Ändere gern die Zahlen und führe die Zelle erneut mit **Shift+Enter** aus, um das neue Ergebnis zu sehen.

Die folgende Diagrammzelle vergleicht die beiden Zahlen und ihre Summe als Balken. So kannst du Werte nicht nur berechnen, sondern auch visuell vergleichen.

````python
zahl_1 = 7
zahl_2 = 5
summe = zahl_1 + zahl_2

print(f"{zahl_1} + {zahl_2} = {summe}")
````

````python
import matplotlib.pyplot as plt

beschriftungen = ["Zahl 1", "Zahl 2", "Summe"]
werte = [zahl_1, zahl_2, summe]

plt.bar(beschriftungen, werte, color=["#20ac27", "#f8e10d", "#f0380a"])
plt.title("Zwei Zahlen und ihre Summe")
plt.ylabel("Wert")
plt.show()
````

## 6. Speichern und Exportieren

Mit **Strg+S** (macOS: **Cmd+S**) speicherst du das Notebook. Die Datei bleibt im Jupyter-Format `.ipynb`; sie enthält Zellen, Code, Text und gegebenenfalls gespeicherte Ausgaben.

Für einen PDF-Export verwende im Jupyter-Menü **File > Save and Export Notebook As > PDF**. Der PDF-Export kann zusätzliche Software wie LaTeX erfordern. In VS Code findest du Exportmöglichkeiten über die Notebook-Befehle bzw. die Befehlspalette.

**Merke:** Wenn du eine frühere Zelle änderst, führe die davon abhängigen späteren Zellen erneut aus. Ein Neustart des Kernels löscht den gespeicherten Python-Zustand; danach müssen die benötigten Codezellen wieder ausgeführt werden. So bleibt das Ergebnis nachvollziehbar.
